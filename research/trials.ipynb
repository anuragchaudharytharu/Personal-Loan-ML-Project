### ensure annnotation

It's is a runtime type-checking decorator.

It enforces that our function arguments match the type hints we wrote.

```python
def get_sum(x:int, y:int) -> int:
    return x * y
get_sum(x = 2, y = 3)
```

If we dont use ensure_annotation, even tho we applied type-checking, we still get result instead of error. This creates massive bug in the code

Python normally does NOT enforce types.

So this will produce:

    "33"

(no error → dangerous in ML pipelines)

So, to force Python to respect our type hints at runtime

We use @ensure_annotations decorator

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_product(x:int, y:int) -> int:
    return x * y
```

```python
get_product(x = 2, y = "3")
```
